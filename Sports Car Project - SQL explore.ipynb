# Sports Car Project - Explore with SQL
In this notebook I will use SQL queries to explore the data. I will import the relelvant modules (pandas, sqlite3), create a connection to the database (`sports_cars.db`) which has an empty table schema already set up. I will then import the data from the `sports_car_clean.csv`. \
The aim of this notebook is to use SQL to:
- View all of the data
- Look at total number of entries, and any null values in each column
- Descriptive statistics for numeric columns
- Descriptive statistics grouped by categories (e.g. grouped by year, engine type and make)
- Look at top 5 cars (i.e. most powerful, most expensive, fastest acceleration)

```python
import sqlite3
import pandas as pd


conn = sqlite3.connect("sports_cars.db")

pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type = 'table';
""", conn)



df = pd.read_csv("sports_car_clean.csv")

df.to_sql(
    "cars",
    conn,
    if_exists="replace",
    index=False
)


pd.options.display.float_format = '{:,.2f}'.format
```

```python
### View the data 
```

```python
pd.read_sql("""
SELECT *
FROM cars;
""", conn)
```

```python
## looking at total number of values and null values in data 
```

```python
pd.read_sql("""
SELECT COUNT(*) AS total_count,
SUM(numeric_engine_size IS NULL) AS null_size,
SUM(engine_type IS NULL) AS null_type,
SUM(horsepower IS NULL) AS null_hp, 
SUM(torque_lb IS NULL) AS null_torque, 
SUM(zero_to_sixty IS NULL) AS null_accel, 
SUM(price_gbp IS NULL) AS null_price
FROM cars;
""", conn)
```

```python
### Descriptive statistics (average, min, max) for each numeric category
```

```python
# engine size 
pd.read_sql("""
SELECT AVG(numeric_engine_size) AS avg_size,
MIN(numeric_engine_size) AS min_size,
MAX(numeric_engine_size) AS max_size
FROM cars;
""", conn)
```

```python
# horsepower 
pd.read_sql("""
SELECT AVG(horsepower) AS avg_hp,
MIN(horsepower) AS min_hp,
MAX(horsepower) AS max_hp
FROM cars;
""", conn)
```

```python
# torque 
pd.read_sql("""
SELECT AVG(torque_lb) AS avg_torque,
MIN(torque_lb) AS min_torque,
MAX(torque_lb) AS max_torque
FROM cars;
""", conn)
```

```python
# 0-60 accel 
pd.read_sql("""
SELECT AVG(zero_to_sixty) AS avg_zero_sixty,
MIN(zero_to_sixty) AS min_zero_sixty,
MAX(zero_to_sixty) AS max_zero_sixty
FROM cars;
""", conn)
```

```python
# price
 
pd.read_sql("""
SELECT AVG(price_gbp) AS avg_price,
MIN(price_gbp) AS min_price,
MAX(price_gbp) AS max_price
FROM cars;
""", conn)
```

```python
### Average grouped by Engine type 
```

```python
pd.read_sql(""" 
SELECT engine_type, 
AVG(horsepower) AS avg_hp, 
AVG(torque_lb) AS avg_torque,
AVG(zero_to_sixty) AS avg_zero_to_sixty,
AVG(price_gbp) AS avg_price
FROM cars
GROUP BY engine_type;
""", conn)
```

```python
### Average grouped by year 
```

```python
pd.read_sql(""" 
SELECT year, 
AVG(horsepower) AS avg_hp, 
AVG(torque_lb) AS avg_torque,
AVG(zero_to_sixty) AS avg_zero_to_sixty,
AVG(price_gbp) AS avg_price
FROM cars
GROUP BY year
ORDER by year;
""", conn)
```

```python
### Average grouped by make (only those with count of more than 10)
```

```python
pd.read_sql(""" 
SELECT make, 
COUNT(*) AS make_count,
AVG(horsepower) AS avg_hp, 
AVG(torque_lb) AS avg_torque,
AVG(zero_to_sixty) AS avg_zero_to_sixty,
AVG(price_gbp) AS avg_price
FROM cars
GROUP BY make
HAVING make_count >= 10
ORDER BY make_count DESC;
""", conn)
```

```python
# top 5 fastest cars (0-60)

pd.read_sql(""" 
SELECT make, 
model,
zero_to_sixty 
FROM cars
ORDER BY zero_to_sixty ASC
LIMIT 5;
""", conn)
```

```python
# top 5 most powerful cars (hp)

pd.read_sql(""" 
SELECT make, model,
horsepower 
FROM cars
ORDER BY horsepower DESC
LIMIT 5;
""", conn)
```

```python
# top 5 most expensive cars 
pd.read_sql(""" 
SELECT make, model,
price_gbp 
FROM cars
ORDER BY price_gbp DESC
LIMIT 5;
""", conn)
```
